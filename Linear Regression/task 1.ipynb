```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,mean_squared_error, mean_absolute_error, r2_score
```

```python
df = pd.read_csv("Data.csv")
df
```

```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
plt.scatter(X, y)
plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
# Split data into train and testing set
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
df = pd.DataFrame({"Actual Value":y_test, "Prediction value":y_pred})
df
```

```python
# model evaluation
#acc = accuracy_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mbr = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
#print("Accuracy Score",acc)
print("Mean Absolute Error",mse)
print("Mean Absolute Error", mbr)
print("r2_score",r2)
```

```python
new_salary = model.predict([[1.3]])
print("Predicted Salary : ", new_salary)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(X, y, color="blue", label="Actual Data")
plt.plot(X, model.predict(X), color="red", linewidth=2, label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,6))

# Actual values
plt.scatter(X_test, y_test, color='blue', label='Actual Salary')

# Predicted values
plt.scatter(X_test, y_pred, color='red', marker='x', s=100, label='Predicted Salary')

plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Actual vs Predicted Salary")
plt.legend()
plt.grid(True)

plt.show()
```
