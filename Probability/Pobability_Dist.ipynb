```python
# Binomial Distrbution
from scipy.stats import binom
n = 5
k = 3
p = 0.5
prob = binom.pmf(k, n, p)
print(prob)
```

```python
# uniform distribution
import numpy as np
import matplotlib.pyplot as plt

#values from 0 to 10
values = np.random.uniform(0, 10, 10_00_000)

#ploting a histogram
plt.hist(values, bins = 100, density = True, alpha = 0.3)
plt.title("Continuous uniform distribution from 0 to 10")
plt.xlabel("X")
plt.ylabel("Probability Desnity")
plt.grid(True)

plt.show()
```

```python
# normal distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

#parameter
mu = 70
sigma = 10

# x - axis values
x = np.linspace(30, 110, 1000)

# normal pdf
y = norm.pdf(x, mu, sigma)

#plot the bell curve
plt.plot(x, y, color = "black", linewidth = 2, label = "Normal Distribution")

# shade 68% region 
x1 = np.linspace(mu - sigma, mu + sigma, 1000)
plt.fill_between(x1, norm.pdf(x1, mu, sigma), color = "green", alpha = 0.3, label = "68% region")

# shade of 95% region
x2 = np.linspace(mu - 2*sigma, mu + 2*sigma, 1000)
plt.fill_between(x2, norm.pdf(x2, mu, sigma), color = "yellow", alpha = 0.2, label = "90% region")

# shade of 99.7% region
x3 = np.linspace(mu - 3*sigma, mu + 3*sigma, 1000)
plt.fill_between(x3, norm.pdf(x3, mu, sigma), color = "red", alpha = 0.1, label = "99.7% region")

plt.title("Normal Distribution")
plt.xlabel("Score")
plt.ylabel("probability Density")
plt.legend()
plt.show()
```
