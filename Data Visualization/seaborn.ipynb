```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
#relational plots

sns.relplot(
    data=tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",
    style = "smoker",
    size = "size"
)
```

```python
#line plot

sns.relplot(
    data=tips,
    x = "total_bill",
    y = "tip",
    kind = "line"
)
```

```python
x_val = [1,2,3,4,5,6,7,8,9]
x_sq = [i**2 for i in x_val]
sns.set_theme(style = "ticks")
sns.relplot(
    x = x_val,
    y = x_sq,
    kind = "line"
)
```

```python
# scatter plot
tips = sns.load_dataset("tips")
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "time"
)
```

```python
# line_plot
flight = sns.load_dataset("flights")
sns.lineplot(
    data = flight,
    x = "year",
    y = "passengers"
)
```

```python
#categorical plots
#1. bar_plot
print(tips.head())
```

```python
sns.barplot(
    data=tips,
    x = "day",
    y ="tip",
    hue = "sex"
)
```

```python
# box_plot
sns.boxplot(
    data = tips,
    x = "day",
    y = "tip"
)
```

```python
#Distribution plot
# histogram
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(
    data=penguins,
    x = "body_mass_g",
    bins = 20
)
```

```python
print(flight.head())
```

```python
flight_pivot = flight.pivot(index = "month", columns = "year", values = "passengers")
sns.heatmap(
    flight_pivot,
    cmap = "coolwarm",
    annot = True,
    fmt  = "d"
)
plt.title("Passenger Heatmap")
plt.tight_layout()
```

```python
fig, axes = plt.subplots()

sns.lineplot(
    data =tips,
    x = "day",
    y = "total_bill",
    hue = "sex",
    marker = "o",
    errorbar=None
)
plt.title("avg bill values for each day")
plt.xlabel("Day")
plt.ylabel("Total_bill")
plt.tight_layout()
```
