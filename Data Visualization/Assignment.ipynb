```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.scatterplot(
    data = penguins,
    x = "flipper_length_mm",
    y = "body_mass_g",
   hue = "species"
)
plt.title("Penguins: Flipper Length vs Body Mass")
plt.show()
```

```python
tips = sns.load_dataset("tips")
# Create box plot
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")

# Add title and labels
plt.title("Total Bill Distribution by Day and Sex")
plt.xlabel("Day")
plt.ylabel("Total Bill")

# Display plot
plt.show()
```

```python
median_bill = tips.groupby("day", observed=True)["total_bill"].median()

print(median_bill)
print("Day with highest median bill:", median_bill.idxmax())
```

```python
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "sex",
    marker= "+",
    style = "smoker"
)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

# Display plot
plt.show()
plt.tight_layout()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset
tips = sns.load_dataset("tips")

# Create pivot table
pivot_table = pd.pivot_table(
    tips,
    index="day",
    columns="time",
    values="tip",
    aggfunc="mean"
)

print(pivot_table)

# Create heatmap
sns.heatmap(
    pivot_table,
    cmap="coolwarm",
    annot=True,
    fmt=".2f"
)

plt.title("Tips Heatmap")
plt.xlabel("Time")
plt.ylabel("Day")
plt.tight_layout()
plt.show()

# Find the highest average tip
print("Best day-time combination:", pivot_table.stack().idxmax())
print("Highest average tip:", round(pivot_table.stack().max(), 2))
```
