```python
import pandas as pd

# Real data
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1065, 170, 427, 133, 232]

# Create DataFrame
df = pd.DataFrame({
    "Movie": oscar_movies,
    "Year": oscar_years,
    "Revenue": oscar_revenue
})

df
```

```python
# plot
plt.plot(oscar_years, oscar_revenue)
plt.title("Oscar Movie Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
```

```python
non_oscar_movies = [
    "Radhe Radhe",
    "Mission Impossible",
    "Django Unchained",
    "If you can then catch me",
    "liger"
]
non_oscar_revenue = [370, 2788, 829, 185, 275]

plt.plot(oscar_years, oscar_revenue, label = "Oscar Movies")
plt.plot(oscar_years, non_oscar_revenue, label = "Non Oscar Movies")
plt.title("Oscar Movies Revenue vs Non Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")
plt.legend()
```

```python
# format string : Consist of a part for colors, markers, line
plt.plot(oscar_years, oscar_revenue, color = "r", linewidth =5, linestyle = "-", marker="o", label="Oscar Movies")
plt.plot(oscar_years, non_oscar_revenue, marker="<", label="Non Oscar Movies")

plt.title("Oscar Movies Revenue vs Non Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")
plt.legend()
plt.grid()

plt.show()
```

```python
# other Attributes
plt.plot(oscar_years, oscar_revenue,  label="Oscar Movies")
plt.plot(oscar_years, non_oscar_revenue, label="Non Oscar Movies")

plt.title("Oscar Movies Revenue vs Non Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")
plt.legend()


print(plt.style.available)
print(plt.style.use("ggplot"))
plt.show()
plt.savefig("Final_chart.png")
```

```python
# bar chart: It display categorical data using rectangular data
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1065, 170, 427, 133, 232]
plt.bar(oscar_years, oscar_revenue)
plt.title("Oscar Movie Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")

for i in range(len(oscar_years)):
    plt.text(oscar_years[i], oscar_revenue[i]+20, str(oscar_revenue[i]), ha = "center")
plt.ylim(0, max(oscar_revenue)+100)
```

```python
import numpy as np

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1065, 170, 427, 133, 232]

non_oscar_movies = [
    "Radhe Radhe",
    "Mission Impossible",
    "Django Unchained",
    "If you can then catch me",
    "liger"
]
non_oscar_revenue = [370, 2788, 829, 185, 275]
x = np.arange(len(oscar_years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label = "Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width, label = "Non-Oscar Movies")
plt.title("Oscar vs Non-Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue")
plt.legend()
```

```python
# scatter plot
people = ["A", "B", "C", "D","E",
          "F", "G", "H", "I", "J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure=[110, 115, 120, 122, 125, 130, 135, 123, 140 , 145]

color = ["green" if x<135 else "red" for x in blood_pressure]

plt.title("Age vs Blood_Pressure")
plt.scatter(age, blood_pressure, color=color, s = blood_pressure, marker = "o", alpha = 0.5)
plt.xlabel(age)
plt.ylabel(blood_pressure)
plt.grid()
```

```python
# scatter plot
people = ["A", "B", "C", "D","E",
          "F", "G", "H", "I", "J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure=[110, 115, 120, 122, 125, 130, 135, 123, 140 , 145]

color = ["green" if x<135 else "red" for x in blood_pressure]

plt.title("Age vs Blood_Pressure")
plt.scatter(age, blood_pressure, cmap="OrRd", c = blood_pressure, s = blood_pressure)
plt.xlabel(age)
plt.ylabel(blood_pressure)
plt.grid()
plt.colorbar(label = "BP")

for i in range(len(people)):
    plt.annotate(people[i], xy = (age[i], blood_pressure[i]), xytext = (age[i]+1, blood_pressure[i]+1))

```
