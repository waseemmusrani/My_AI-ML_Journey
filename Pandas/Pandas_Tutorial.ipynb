# DataFrame
- DataFrame Pandas ka 2-dimensional data structure hai jo data ko rows aur columns ki form mein store karta hai.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Usman"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 92]
})
print(df)
```

```python
# DataFrame Methods

print("----- head() -----")
print(df.head()) 
# head() displays the first 5 rows of the DataFrame

print("\n----- head(2) -----")
print(df.head(2)) 
# head(2) displays the first 2 rows of the DataFrame

print("\n----- tail() -----")
print(df.tail()) 
# tail() displays the last 5 rows of the DataFrame

print("\n----- info() -----")
df.info()
# info() displays basic information about the DataFrame

print("\n----- describe() -----")
print(df.describe())
# describe() provides statistical summary of numerical columns

print("\n----- shape -----")
print(df.shape)
# shape returns the number of rows and columns in the DataFrame

print("\n----- columns -----")
print(df.columns)
# columns returns the names of all columns

print("\n----- dtypes -----")
print(df.dtypes)
# dtypes shows the data type of each column

print("\n----- sort_values() Ascending -----")
print(df.sort_values("Marks"))
# sort_values() sorts the DataFrame by a column in ascending order

print("\n----- sort_values() Descending -----")
print(df.sort_values("Marks", ascending=False))
# sort_values() sorts the DataFrame by a column in descending order

print("\n----- drop() Column -----")
print(df.drop("Age", axis=1))
# drop() removes a column from the DataFrame

print("\n----- drop() Row -----")
print(df.drop(0, axis=0))
# drop() removes a row from the DataFrame

print("\n----- dropna() -----")
print(df.dropna())
# dropna() removes rows or columns containing missing values

print("\n----- fillna() -----")
print(df.fillna(0))
# fillna() replaces missing values with a specified value

print("\n----- isnull() -----")
print(df.isnull())
# isnull() checks whether the DataFrame contains missing values

print("\n----- isnull().sum() -----")
print(df.isnull().sum())
# isnull().sum() counts the total number of missing values in each column

print("\n----- nunique() -----")
print(df["Age"].nunique())
# nunique() returns the number of unique values in a column

print("\n----- duplicated() -----")
print(df.duplicated())
# duplicated() checks whether rows are duplicated

print("\n----- drop_duplicates() -----")
print(df.drop_duplicates())
# drop_duplicates() removes duplicate rows from the DataFrame

print("\n----- set_index() -----")
print(df.set_index("Name"))
# set_index() sets a specified column as the DataFrame index

print("\n----- reset_index() -----")
print(df.reset_index())
# reset_index() resets the index back to the default numeric index
```

```python
#melt function = wide to long format
import pandas as pd

df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})
melted_df = df.melt(
    id_vars = ["country", "year"],
    value_vars = ["sales", "profit"],
    var_name = "metrics",
    value_name= "value"
)
melted_df
```

```python
# pivot = long to wide format
original = melted_df.pivot(
    index = ["country", "year"],
    columns = "metrics",
    values = "value"
)
original
```

```python
# Merging and joining data

import pandas as pd

# Customers DataFrame
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

# Orders DataFrame
df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

# Merge both DataFrames
pd.merge(df_customers, df_orders, on = "customer_id") #using inner join
```

```python
pd.merge(df_customers, df_orders, on = "customer_id", how = "left") #using inner join
```

```python
pd.merge(df_customers, df_orders, on = "customer_id", how = "right") #using inner join
```

```python
#Data Concatenation
#Stacking datafame on the top of each other(rows) or side by side(columns)
df1 = pd.DataFrame({
    "id" : [1, 2, 3],
    "name" : ["adam", "eve", "bob"]
})

df2 = pd.DataFrame({
    "id" : [4, 5, 6],
    "name" : ["charlie", "joy", "coob"]
})

pd.concat([df2, df1]) #row wise concatenation
```

```python
pd.concat([df2, df1], axis = 1) #column wise concatenation
```
