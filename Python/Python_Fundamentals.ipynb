````python
#paindrome

def palindrome(number):

    original = number
    reverse = 0

    while number > 0:
        digit = number % 10
        reverse = reverse * 10 + digit
        number //= 10

    return original == reverse

num = int(input("Enter number: "))

if palindrome(num):
    print("Palindrome")
else:
    print("Not Palindrome")
````

````python
def armstrong(number):

    total = 0
    temp = number

    while temp > 0:
        digit = temp % 10
        total += digit ** len(str(number))
        temp //= 10

    return total == number

num = int(input("Enter number: "))

if armstrong(num):
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
    
````

````python
def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n-1)

print(factorial(5))
````

# 6. Encapsulation

## Definition
Encapsulation ka matlab hai **data aur methods ko ek class ke andar bundle karna** aur sensitive data ko controlled access dena.

## Simple Explanation
Sensitive data ko direct access dene ke bajaye methods ke through controlled way mein use kiya jata hai.

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
class ClassName:
    def __init__(self):
        self.__private_data = value

    def method(self):
        # controlled access
        pass
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient Balance")

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(1500)
account.deposit(500)
account.withdraw(300)
account.show_balance()
````

## Output
```text
Balance: 1700
```

## Key Points
- Data aur methods ko ek class mein bundle kiya jata hai.
- `__balance` private-style attribute hai.
- Data ko methods ke through controlled way mein access kiya jata hai.

# 7. Inheritance

## Definition
Inheritance OOP ka concept hai jisme ek class doosri class ke attributes aur methods ko inherit/reuse kar sakti hai.

## Simple Explanation
Child class parent class ki properties aur methods ko use kar sakti hai. Isse code reuse hota hai.

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
class Parent:
    pass

class Child(Parent):
    pass
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
class Animal:
    def eat(self):
        print("Animal is eating")

    def sleep(self):
        print("Animal is sleeping")


class Dog(Animal):
    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()
dog.sleep()
dog.bark()
````

## Output
```text
Animal is eating
Animal is sleeping
Dog is barking
```

## Key Points
- Parent class ke features child class mein reuse ho sakte hain.
- Inheritance code reusability improve karti hai.
- Child class apne additional methods bhi define kar sakti hai.

# 8. `super()`

## Definition
`super()` ka use parent class ke methods ya constructor ko call karne ke liye hota hai.

## Simple Explanation
Agar child class apne constructor ke saath parent class ka constructor bhi use karna chahe, to `super()` helpful hota hai.

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
class Child(Parent):
    def __init__(self, value):
        super().__init__(value)
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

    def show_roll_no(self):
        print("Roll No:", self.roll_no)


student = Student("Waseem", 101)

student.show_name()
student.show_roll_no()
````

## Output
```text
Name: Waseem
Roll No: 101
```

## Key Points
- Parent constructor ko call karne mein commonly use hota hai.
- Inherited code ko reuse karna easy hota hai.
- Child class parent ki functionality ko extend kar sakti hai.

# 9. Multiple Inheritance

Multiple inheritance mein ek child class **multiple parent classes** se inherit karti hai.

## Simple Explanation
Ek class ek se zyada parent classes ke methods use kar sakti hai.

# 10. Multilevel Inheritance

## Definition
Multilevel inheritance mein inheritance multiple levels mein hoti hai.

## Simple Explanation
Ek child class kisi parent class se inherit karti hai, aur phir next class us child class se inherit karti hai.

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
class A:
    pass

class B(A):
    pass

class C(B):
    pass
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no

    def show_roll_no(self):
        print("Roll No:", self.roll_no)


class UniversityStudent(Student):
    def __init__(self, name, roll_no, university):
        super().__init__(name, roll_no)
        self.university = university

    def show_university(self):
        print("University:", self.university)


student = UniversityStudent("Waseem", 34, "FUUAST")

student.show_name()
student.show_roll_no()
student.show_university()
````

## Output
```text
Name: Waseem
Roll No: 34
University: FUUAST
```

## Key Points
- Inheritance multiple levels mein ho sakti hai.
- Last-level child ko upper-level classes ke accessible features milte hain.
- `super()` parent level ke constructor ko call karne mein use hota hai.

# 11. Abstraction

## Definition
Abstraction OOP ka concept hai jisme unnecessary/internal details ko hide kiya jata hai aur user ko sirf important functionality show ki jati hai.

## Simple Explanation
**"How kaam ho raha hai" ko hide karo, "kya kaam karna hai" ko show karo.**

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
from abc import ABC, abstractmethod

class Parent(ABC):
    @abstractmethod
    def method(self):
        pass
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass


class Dog(Animal):
    def make_sound(self):
        print("Dog is barking")


class Cat(Animal):
    def make_sound(self):
        print("Meow meow")


dog = Dog()
cat = Cat()

dog.make_sound()
cat.make_sound()
````

## Output
```text
Dog is barking
Meow meow
```

## Key Points
- Unnecessary/internal details hide ki ja sakti hain.
- Abstract class mein abstract methods define kiye ja sakte hain.
- Child classes abstract methods ki implementation provide karti hain.

# 13. Duck Typing

## Definition
Duck typing mein Python exact object type se zyada object ke required behavior/method ko importance deta hai.

## Simple Explanation
Agar object ke paas required method hai, to function us object ko use kar sakta hai. Python mein behavior important hota hai.

## Syntax
Neeche basic syntax executable **Code Cell** mein diya gaya hai.

````python
def function(object):
    object.required_method()
````

## Practical Example
Neeche practical example executable **Code Cell** mein diya gaya hai.

````python
class Dog:
    def sound(self):
        print("Dog says Woof")


class Cat:
    def sound(self):
        print("Cat says Meow")


def make_sound(animal):
    animal.sound()


dog = Dog()
cat = Cat()

make_sound(dog)
make_sound(cat)
````

## Output
```text
Dog says Woof
Cat says Meow
```

## Key Points
- Exact object type check karna zaroori nahi hota.
- Required method/behavior available hona important hai.
- Duck typing Python ke flexible style ko demonstrate karta hai.
